# Training
- Configuration possibilities:

| Flag name | Type | Help string | Default value | Choices |
|------------|------|-------------|---------------|---------|
| `--path` | `str` | custom path | None | — |
| `--data` | `str` | dataset name | None | — |
| `--config` | `str` | path to config file | None | — |
| `--gpu` | `str` | which GPU to use | `"0"` | — |
| `--model_name` | `str` | name of stored model | `""` | — |
| `--base_model` | `str` | base model structure | `"origin"` | `origin`, `np`, `snp`, `anp`, `mnp` |
| `--ode` | `bool` (`store_true`) | enable ODE decoder | `False` | — |
| `-d`, `--determinstic` | `bool` (`store_true`) | enable determinstic path decoder | `False` | — |
| `--resize_ratio` | `int` | timestamp resize ratio | `0` | — |
| `--rand_edge_features` | `int` | use random edge featrues | `0` | — |
| `--rand_node_features` | `int` | use random node featrues | `0` | — |
| `--eval_neg_samples` | `int` | how many negative samples to use at inference. Note: this will change the metric of test set to AP+AUC to AP+MRR! | `1` | — |
| `--zero_feat` | `bool` (`store_true`) | use all 0 featrues | `False` | — |
| `--train_ratio` | `float` | precentage of training ratio | `0` | — |
| `--seed` | `int` | random seed | `123` | — |
| `--latent_dist` | `str` | latent distribution used by the neural process: gaussian or gmm | `"gmm"` | `gaussian`, `gmm` |
| `--n_components` | `int` | Number of components in the Gaussian Mixture Model | `3` | — |
